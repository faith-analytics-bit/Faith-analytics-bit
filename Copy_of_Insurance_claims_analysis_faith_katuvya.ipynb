```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create synthetic claims data - 1000 claims
np.random.seed(42)
n = 1000
data = {
    'Claim_ID': range(1, n+1),
    'Age': np.random.randint(20, 70, n),
    'Gender': np.random.choice(['Male','Female'], n),
    'Region': np.random.choice(['Nairobi','Mombasa','Kisumu','Nakuru','Kitale'], n),
    'Claim_Type': np.random.choice(['Motor','Medical','Life','Property'], n),
    'Claim_Amount_KES': np.random.randint(5000, 500000, n)
}
df = pd.DataFrame(data)

# 1. Total claims & average
print("Total Claims:", len(df))
print("Total Paid (KES):", df['Claim_Amount_KES'].sum())
print("Average Claim (KES):", round(df['Claim_Amount_KES'].mean(), 2))

# 2. Which Claim Type costs most?
print("\nBy Claim Type:")
print(df.groupby('Claim_Type')['Claim_Amount_KES'].agg(['count','mean','sum']).round(2))

# 3. Which Region costs most?
print("\nBy Region:")
print(df.groupby('Region')['Claim_Amount_KES'].agg(['count','mean','sum']).round(2))

# 4. Chart - Average Claim by Type
df.groupby('Claim_Type')['Claim_Amount_KES'].mean().plot(kind='bar', color='skyblue')
plt.title('Average Claim Amount by Claim Type')
plt.ylabel('KES')
plt.show()
```
